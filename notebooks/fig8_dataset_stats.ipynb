```python
import matplotlib.pyplot as plt
import collections
import statistics
import os
import glob
import numpy as np
from matplotlib.ticker import MaxNLocator  

plt.rcParams['text.usetex'] = False 

# Use Computer Modern for math text

plt.rcParams['mathtext.fontset'] = 'cm'


plt.rcParams['font.family'] = 'serif'

def plot_subclass_distribution(counts_per_file):
    """
    Bar chart of subclass counts.
    """
    
    plt.rcParams['font.family'] = 'sans-serif'
    plt.rcParams['font.sans-serif'] = ['Arial', 'Helvetica']
    plt.rcParams['font.size'] = 12
    plt.rcParams['axes.linewidth'] = 1.2

    if not counts_per_file:
        print("no data")
        return

    
    counter = collections.Counter(counts_per_file)
    
    min_x = min(counter.keys())
    max_x = max(counter.keys())
    
    all_x_values = list(range(min_x, max_x + 1))
    all_y_values = [counter.get(x, 0) for x in all_x_values]

    
    fig, ax = plt.subplots(figsize=(4.8, 3.0), dpi=150)

    
    bars = ax.bar(all_x_values, all_y_values, 
                  color='#CFEAF1', 
                  edgecolor='black', 
                  linewidth=1.2, 
                  alpha=1.0, 
                  width=0.8, 
                  zorder=3)

    
    for bar in bars:
        height = bar.get_height()
        if height > 0:
            ax.text(bar.get_x() + bar.get_width()/2., height + 0.1,
                     f'{int(height)}',
                     ha='center', va='bottom', 
                     fontsize=13, fontweight='bold', color='black')

    
    ax.set_xlabel('Number of Derived Classes', fontsize=13, fontweight='bold', labelpad=5)
    ax.set_ylabel('Number of Tool Templates', fontsize=13, fontweight='bold', labelpad=5)

    
    ax.set_xticks(all_x_values)
    
    # Force integer Y-axis ticks
    ax.yaxis.set_major_locator(MaxNLocator(integer=True))
    
    
    max_y_data = max(all_y_values)
    
    top_margin = 1.15 if max_y_data > 2 else 1.5
    ax.set_ylim(0, max_y_data * top_margin) 

    ax.tick_params(axis='both', which='major', labelsize=12)

    ax.grid(False) 
    ax.grid(axis='y', linestyle='--', color='gray', alpha=0.5, zorder=0) 

    
    for spine in ax.spines.values():
        spine.set_visible(True)
        spine.set_edgecolor('black')
        spine.set_linewidth(1.5)

    plt.tight_layout()

    plt.savefig('figs/app/data_dervied_class.pdf', bbox_inches='tight', dpi=300)
    plt.show()

# ---------------------------------------------------------

# ---------------------------------------------------------
def analyze_and_plot_subclasses(directory_path):
    counts_per_file = []
    
    json_files = glob.glob(os.path.join(directory_path, "*.json"))
    
    if not json_files:
        print(f"no JSON found in {directory_path}")
        return

    for file_path in json_files:
        try:
            with open(file_path, 'r', encoding='utf-8') as f:
                data = json.load(f)
                meta_data = data.get("meta_data", {})
                subclasses = meta_data.get("derived_classes", [])
                
                if isinstance(subclasses, list):
                    counts_per_file.append(len(subclasses))
        except Exception:
            pass 

    if not counts_per_file:
        return

    print(f"mean: {statistics.mean(counts_per_file):.2f}, max: {max(counts_per_file)}, min: {min(counts_per_file)}")
    
    plot_subclass_distribution(counts_per_file)


target_directory = "data/SABEval/tool_templates"
if os.path.exists(target_directory):
    analyze_and_plot_subclasses(target_directory)
```

```python
import os
import json
import glob
import statistics

def analyze_properties_count(directory_path):
    counts_per_file = []
    
    
    json_files = glob.glob(os.path.join(directory_path, "*.json"))
    
    if not json_files:
        print(f"no JSON found in {directory_path}")
        return

    for file_path in json_files:
        try:
            with open(file_path, 'r', encoding='utf-8') as f:
                data = json.load(f)
                
                # navigate tool_template -> parameters -> properties
                tool_template = data.get("tool_template", {})
                parameters = tool_template.get("parameters", {})
                properties = parameters.get("properties", {})
                
                
                if isinstance(properties, dict):
                    count = len(properties.keys())
                    counts_per_file.append(count)
                    
                    # print(f"{os.path.basename(file_path)}: {count}")
                    
        except Exception as e:
            print(f"error processing {file_path}: {e}")
            pass 

    if not counts_per_file:
        print("no JSON with valid properties data")
        return

    print(f"files: {len(counts_per_file)}")
    print(f"mean: {statistics.mean(counts_per_file):.2f}, max: {max(counts_per_file)}, min: {min(counts_per_file)}")
    

target_directory = "data/SABEval/tool_templates"
if os.path.exists(target_directory):
    analyze_properties_count(target_directory)
else:
    print(f"directory not found: {target_directory}")
```

```python
import os
import json
import glob

def analyze_unique_subclasses(directory_path):
    all_subclasses = set() # use a set to deduplicate
    
    
    json_files = glob.glob(os.path.join(directory_path, "*.json"))
    
    if not json_files:
        print(f"no JSON found in {directory_path}")
        return

    for file_path in json_files:
        try:
            with open(file_path, 'r', encoding='utf-8') as f:
                data = json.load(f)
                
                
                meta_data = data.get("meta_data", {})
                subclasses = meta_data.get("derived_classes", [])
                
                
                if isinstance(subclasses, list):
                    for sub in subclasses:
                        if isinstance(sub, str):
                            all_subclasses.add(sub.strip()) 
                            
        except Exception as e:
            print(f"error processing {file_path}: {e}")
            pass 

    if not all_subclasses:
        print("no subclass data")
        return

    print(f"done.")
    print(f"distinct subclasses: {len(all_subclasses)}")
    
    
    # print(f"items: {list(all_subclasses)}")
    

target_directory = "data/SABEval/tool_templates"
if os.path.exists(target_directory):
    analyze_unique_subclasses(target_directory)
else:
    print(f"directory not found: {target_directory}")
```
